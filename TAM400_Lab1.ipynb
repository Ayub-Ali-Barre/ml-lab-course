## 4. Introduction to Machine Learning Workflow

A typical Machine Learning pipeline consists of the following 5 key steps:
1. **Data Collection:** Gathering the raw data required to solve the problem.
2. **Data Preprocessing:** Cleaning missing values, handling outliers, encoding categoricals, and normalization.
3. **Feature Engineering:** Selecting and transforming variables to improve model performance.
4. **Model Training:** Fitting a machine learning algorithm (e.g., Linear Regression) on the training set.
5. **Model Evaluation:** Assessing the model using test metrics (e.g., MSE, $R^2$, Accuracy).

````python
# 4. Machine Learning Workflow Example: Linear Regression
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np
import matplotlib.pyplot as plt

# 1. Data Collection (Generating synthetic sample data)
np.random.seed(42)
X_train = np.linspace(0, 10, 50).reshape(-1, 1)
y_train = 2.5 * X_train.flatten() + 4.0 + np.random.normal(0, 2.0, size=50)

# 4. Model Training (as shown in Lab guide)
model = LinearRegression()
model.fit(X_train, y_train)

# 5. Model Evaluation
y_pred = model.predict(X_train)
print(f"Learned Slope (w): {model.coef_[0]:.4f}")
print(f"Learned Intercept (b): {model.intercept_:.4f}")
print(f"Mean Squared Error (MSE): {mean_squared_error(y_train, y_pred):.4f}")
print(f"R^2 Score: {r2_score(y_train, y_pred):.4f}")

# Plotting the result
plt.figure(figsize=(8, 5))
plt.scatter(X_train, y_train, color="royalblue", label="Data Points", alpha=0.7)
plt.plot(X_train, y_pred, color="crimson", linewidth=2, label="Fitted Linear Model")
plt.title("ML Pipeline Example: Linear Regression")
plt.xlabel("X (Feature)")
plt.ylabel("y (Target)")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
````

## 5. Search Algorithms in Python

Search algorithms are fundamental in Artificial Intelligence. They are used to explore possible solutions and find a path from a start node to a goal node.

In this section, we implement three basic search algorithms:
* **Breadth-First Search (BFS)**
* **Depth-First Search (DFS)**
* **A* Search Algorithm**

---

### 5.1 Problem Setup
We represent a graph where nodes are connected to their neighbors using an adjacency list (Python dictionary):

```
        A
       / \
      B   C
     / \   \
    D   E   F
         \ /
          F
```

````python
# 5.1 Problem Setup: Graph Representation
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

print("Graph structure:")
for node, neighbors in graph.items():
    print(f"  Node {node} -> {neighbors}")
````

---
### 5.2 Breadth-First Search (BFS)

**BFS explores nodes level by level** using a First-In-First-Out (**FIFO**) queue (`collections.deque`).
* It explores all immediate neighbors of the current depth before moving to the next level.
* In unweighted graphs, BFS is guaranteed to find the shortest path in terms of the number of edges/hops.

````python
# 5.2 Breadth-First Search (BFS) Implementation
from collections import deque

def bfs(graph, start, goal):
    queue = deque([[start]])
    visited = set()

    while queue:
        path = queue.popleft()
        node = path[-1]

        if node == goal:
            return path

        if node not in visited:
            visited.add(node)

            for neighbor in graph[node]:
                new_path = list(path)
                new_path.append(neighbor)
                queue.append(new_path)

    return None

print("BFS Path:", bfs(graph, 'A', 'F'))
````

---
### 5.3 Depth-First Search (DFS)

**DFS explores as deep as possible before backtracking** using a Last-In-First-Out (**LIFO**) stack (a standard Python list).
* It follows a branch until it hits a dead end or reaches the goal node, then backtracks.
* Memory efficient, but not guaranteed to find the shortest path.

````python
# 5.3 Depth-First Search (DFS) Implementation
def dfs(graph, start, goal):
    stack = [[start]]
    visited = set()

    while stack:
        path = stack.pop()
        node = path[-1]

        if node == goal:
            return path

        if node not in visited:
            visited.add(node)

            for neighbor in graph[node]:
                new_path = list(path)
                new_path.append(neighbor)
                stack.append(new_path)

    return None

print("DFS Path:", dfs(graph, 'A', 'F'))
````

---
### 5.4 A* Search Algorithm

**A* Search** uses both a path cost function $g(n)$ and a heuristic function $h(n)$ to find the optimal path:
$$f(n) = g(n) + h(n)$$
* $g(n)$: The exact cost incurred from the start node to node $n$.
* $h(n)$: The estimated (heuristic) cost from node $n$ to the goal.
* $f(n)$: Total estimated cost of path through node $n$.

A priority queue (`heapq`) is used to always expand the node with the lowest $f(n)$ value.

#### Example 1 Setup:

````python
# 5.4 A* Search Algorithm - Example 1
import heapq

# Graph with costs (edge weights)
graph_cost = {
    'A': {'B': 1, 'C': 3},
    'B': {'D': 1, 'E': 3},
    'C': {'F': 5},
    'D': {},
    'E': {'F': 1},
    'F': {}
}

# Heuristic values (estimated distance to goal 'F')
heuristic = {
    'A': 5,
    'B': 3,
    'C': 4,
    'D': 2,
    'E': 1,
    'F': 0
}

def a_star(graph, heuristic, start, goal):
    queue = [(0, start, [start])]
    visited = set()

    while queue:
        (cost, node, path) = heapq.heappop(queue)

        if node == goal:
            return path

        if node not in visited:
            visited.add(node)

            for neighbor, weight in graph[node].items():
                new_cost = cost + weight + heuristic[neighbor]
                heapq.heappush(queue, (new_cost, neighbor, path + [neighbor]))

    return None

print("A* Path:", a_star(graph_cost, heuristic, 'A', 'F'))
````

#### A* Search: Another Example (Example 2)
This example demonstrates how A* selects the optimal path using both edge cost and heuristic values more clearly.

```
       (1)       (2)
    S -----> A -------> C
    |                   |
 (4)|                (3)|
    v                   v
    B -----> C -------> G
       (1)       (3)
```

````python
# A* Search Another Example (Example 2)
import heapq

graph = {
    'S': {'A': 1, 'B': 4},
    'A': {'C': 2},
    'B': {'C': 1},
    'C': {'G': 3},
    'G': {}
}

heuristic = {
    'S': 5,
    'A': 4,
    'B': 2,
    'C': 1,
    'G': 0
}

def a_star(graph, heuristic, start, goal):
    queue = [(0, start, [start])]
    visited = set()

    while queue:
        cost, node, path = heapq.heappop(queue)

        if node == goal:
            return path

        if node not in visited:
            visited.add(node)

            for neighbor, weight in graph[node].items():
                new_cost = cost + weight + heuristic[neighbor]
                heapq.heappush(queue, (new_cost, neighbor, path + [neighbor]))

    return None

result = a_star(graph, heuristic, 'S', 'G')
print("A* Path:", result)
````

### 5.6 Hill Climbing (Local Search / Optimization)

**Hill Climbing** is an iterative optimization algorithm that begins with an arbitrary solution and continually moves in the direction of increasing value (uphill) until no neighbor has a higher value.

#### Objective: Maximize the quadratic function:
$$f(x) = -x^2 + 10x$$

* **Analytical Maximum:**
  $$\frac{df}{dx} = -2x + 10 = 0 \implies x^* = 5$$
  $$f(5) = -(5)^2 + 10(5) = -25 + 50 = 25$$

````python
# Visualization of the Objective Function f(x) = -x^2 + 10x
import numpy as np
import matplotlib.pyplot as plt

x_vals = np.linspace(-5, 15, 200)
y_vals = -x_vals**2 + 10 * x_vals

plt.figure(figsize=(8, 4.5))
plt.plot(x_vals, y_vals, label="f(x) = -x^2 + 10x", color="darkviolet", linewidth=2)
plt.scatter([5], [25], color="red", s=90, zorder=5, label="Global Optimum: x=5, f(x)=25")
plt.title("Hill Climbing Function Landscape: f(x) = -x^2 + 10x")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()
plt.show()
````

````python
# 5.6 Hill Climbing Implementation
import random

def f(x):
    return -x**2 + 10*x

def hill_climbing(start, step_size=1, max_iterations=100):
    current = start
    current_value = f(current)

    for i in range(max_iterations):
        neighbors = [current - step_size, current + step_size]

        next_node = current
        next_value = current_value

        for n in neighbors:
            value = f(n)
            if value > next_value:
                next_node = n
                next_value = value

        if next_value <= current_value:
            break

        current = next_node
        current_value = next_value

    return current, current_value

start_point = random.randint(-10, 10)
solution, value = hill_climbing(start_point)

print("Start:", start_point)
print("Best solution x:", solution)
print("Max value f(x):", value)
````

#### Multiple Trials: Do we always get the same result?
Let us test the algorithm across several different starting points.

````python
# Testing Hill Climbing across various starting values
test_starts = [-10, -5, 0, 2, 5, 7, 10]
print(f"{'Start Point':>12} | {'Solution x':>12} | {'Max f(x)':>12}")
print("-" * 42)
for s in test_starts:
    sol, val = hill_climbing(s)
    print(f"{s:>12} | {sol:>12} | {val:>12}")
````

### Analysis Question: "Do we always get the same result?"

**Answer:**
1. **For this specific function $f(x) = -x^2 + 10x$:**
   * **Yes, we always get the same result ($x = 5, f(x) = 25$)**, provided the step size is $1$ and the starting point is an integer.
   * This is because $f(x)$ is a **strictly concave (unimodal) parabola** with a single global maximum at $x=5$ and no local maxima or deceptive plateaus. Every step that increases $f(x)$ moves directly toward $x = 5$.

2. **In general optimization problems:**
   * **No, hill climbing does not always yield the same result.**
   * For complex, multimodal objective functions, hill climbing is greedy and local: it can easily become trapped in **local maxima**, wander on **flat plateaus**, or oscillate along **ridges**. The final solution heavily depends on the initial starting point (`start_point`).
